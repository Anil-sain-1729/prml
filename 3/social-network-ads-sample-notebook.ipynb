```python
# Import necessary libraries
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings('ignore')
```

```python
# Load the dataset
df = pd.read_csv('Social_Network_Ads.csv')
```

```python
# Display the first few rows of the dataset
df.head()
```

```python
# Display shape of the dataset
df.shape
```

```python
# Display column names of the dataset
df.columns
```

```python
# Display summary statistics
df.describe()
```

```python
# Count plot for gender distribution
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='Gender', hue='Purchased', palette='rocket')
plt.title('Purchase Distribution by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Users')
plt.show()
```

```python
# Age distribution for purchased vs non-purchased
plt.figure(figsize=(12, 6))
sns.histplot(data=df, x='Age', hue='Purchased', multiple='stack', palette='rocket', bins=30)
plt.title('Age Distribution by Purchase Status')
plt.xlabel('Age')
plt.ylabel('Number of Users')
plt.show()
```

```python
# Estimated salary distribution for purchased vs non-purchased
plt.figure(figsize=(12, 6))
sns.histplot(data=df, x='EstimatedSalary', hue='Purchased', multiple='stack', palette='rocket', bins=30)
plt.title('Estimated Salary Distribution by Purchase Status')
plt.xlabel('Estimated Salary')
plt.ylabel('Number of Users')
plt.show()
```

```python
# Correlation heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(df[['Age', 'EstimatedSalary', 'Purchased']].corr(), annot=True, annot_kws={"size": 16}, cmap='rocket', vmin=-1, vmax=1)
plt.title('Correlation Matrix')
plt.show()
```
